# V65 - Rank-label targets

Pre-registered plan: NDCG consumes order, not raw scores, and raw relevance is skewed with 71% zeros. Training Ridge on per-user rank labels instead of raw scores may fit the ranking geometry better. Identical word-only features and grouped folds under a fresh seed, gated against raw-score Ridge with the usual locks.

```python
import re
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import rankdata
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import Ridge
from sklearn.model_selection import StratifiedGroupKFold

SEED = 505
N_SPLITS = 5
GATE_FLOOR = 0.002
rng = np.random.RandomState(SEED)

CANDIDATES = [Path("/kaggle/input/intelligo-data/data"), Path("data")]
DATA = next((p for p in CANDIDATES if (p / "train_relevance.csv").exists()), None)
assert DATA is not None, "train_relevance.csv not found"
OUT = Path("/kaggle/working" if Path("/kaggle/working").exists() else "results/v65")
OUT.mkdir(parents=True, exist_ok=True)
print("DATA:", DATA.resolve())
```

## Raw vs rank targets, same features and folds

Rank labels are per-user ranks scaled to unit range. Everything else is held constant so only the target differs.

```python
MODS = [f"M_{i:03d}" for i in range(1, 18)]
train = pd.read_csv(DATA / "train_relevance.csv")
test_ids = pd.read_csv(DATA / "test.csv")["user_id"].tolist()
chats = pd.read_csv(DATA / "chat_history.csv")
assert not (set(train["user_id"]) & set(test_ids))
Y = train[MODS].to_numpy(float)
n_train = len(train)
YR = np.apply_along_axis(lambda r: rankdata(r, method="average") / len(r), 1, Y)
by_user = chats.groupby("user_id")["user_chat_text"].apply(list)
docs = [" ".join(str(t) for t in by_user.get(u, [])) for u in list(train["user_id"])]
W = TfidfVectorizer(max_features=5000, ngram_range=(1, 1), min_df=3).fit_transform(docs).toarray()

def template_group(texts):
    if not texts:
        return "NO_CHAT"
    return re.sub(r"\d+", "#", " ".join(str(texts[0]).lower().split()[:6]))
fold_groups = np.array([template_group(by_user.get(u, [])) for u in train["user_id"]])
strata = Y.argmax(1)

def ndcg5(yt, ys):
    order = np.argsort(-ys, axis=1)
    rel = np.take_along_axis(yt, order, 1)[:, :5]
    disc = 1.0 / np.log2(np.arange(2, 7))
    dcg = (rel * disc).sum(1)
    ideal = np.sort(yt, axis=1)[:, ::-1][:, :5]
    idcg = (ideal * disc).sum(1)
    return np.where(idcg > 0, dcg / np.maximum(idcg, 1e-12), 0.0)

def ridge_oof(T, alpha=1.0):
    oof = np.zeros_like(Y)
    fids = np.zeros(n_train, int)
    skf = StratifiedGroupKFold(n_splits=N_SPLITS, shuffle=True, random_state=SEED)
    for fi, (tr, va) in enumerate(skf.split(W, strata, fold_groups)):
        for j in range(len(MODS)):
            m = Ridge(alpha=alpha)
            m.fit(W[tr], T[tr, j])
            oof[va, j] = m.predict(W[va])
        fids[va] = fi + 1
    return oof, fids

oof_raw, fid = ridge_oof(Y)
oof_rnk, _ = ridge_oof(YR)
base = ndcg5(Y, oof_raw)
d = ndcg5(Y, oof_rnk) - base
lo, hi = np.quantile([rng.choice(d, len(d), replace=True).mean() for _ in range(2000)], [0.025, 0.975])
wf = min(ndcg5(Y[fid == k], oof_rnk[fid == k]).mean() - base[fid == k].mean() for k in range(1, N_SPLITS + 1))
print(f"raw={base.mean():.5f} rank={ndcg5(Y, oof_rnk).mean():.5f} delta={d.mean():+.5f} CI95=[{lo:+.5f},{hi:+.5f}] worst-fold={wf:+.5f}")
print("ADOPT rank targets" if (lo > GATE_FLOOR and wf >= 0) else "KEEP raw targets; rank transform adds nothing")
```
