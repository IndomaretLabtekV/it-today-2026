# V86 - Module co-occurrence boost

Pre-registered plan: labels have structure no feature view uses, which modules co-occur as relevant for the same user. A co-occurrence matrix from training folds boosts the modules that co-occur with the baseline top-1, with fixed weight 0.2 set up front. Same grouped folds under a fresh seed, same locks.

```python
import re
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import Ridge
from sklearn.model_selection import StratifiedGroupKFold

SEED = 1212
N_SPLITS = 5
GATE_FLOOR = 0.002
LAM = 0.2
rng = np.random.RandomState(SEED)

CANDIDATES = [Path("/kaggle/input/intelligo-data/data"), Path("data")]
DATA = next((p for p in CANDIDATES if (p / "train_relevance.csv").exists()), None)
assert DATA is not None, "train_relevance.csv not found"
OUT = Path("/kaggle/working" if Path("/kaggle/working").exists() else "results/v86")
OUT.mkdir(parents=True, exist_ok=True)
print("DATA:", DATA.resolve())
```

## Co-occurrence from train folds only, then gate

The matrix never sees validation labels. Boost applies to the baseline top-1 per user.

```python
MODS = [f"M_{i:03d}" for i in range(1, 18)]
train = pd.read_csv(DATA / "train_relevance.csv")
test_ids = pd.read_csv(DATA / "test.csv")["user_id"].tolist()
chats = pd.read_csv(DATA / "chat_history.csv")
assert not (set(train["user_id"]) & set(test_ids))
Y = train[MODS].to_numpy(float)
n_train = len(train)
by_user = chats.groupby("user_id")["user_chat_text"].apply(list)
docs = [" ".join(str(t) for t in by_user.get(u, [])) for u in list(train["user_id"])]
W = TfidfVectorizer(max_features=5000, ngram_range=(1, 1), min_df=3).fit_transform(docs).toarray()

def template_group(texts):
    if not texts:
        return "NO_CHAT"
    return re.sub(r"\d+", "#", " ".join(str(texts[0]).lower().split()[:6]))
fold_groups = np.array([template_group(by_user.get(u, [])) for u in train["user_id"]])
strata = Y.argmax(1)

def ndcg5(yt, ys):
    order = np.argsort(-ys, axis=1)
    rel = np.take_along_axis(yt, order, 1)[:, :5]
    disc = 1.0 / np.log2(np.arange(2, 7))
    dcg = (rel * disc).sum(1)
    ideal = np.sort(yt, axis=1)[:, ::-1][:, :5]
    idcg = (ideal * disc).sum(1)
    return np.where(idcg > 0, dcg / np.maximum(idcg, 1e-12), 0.0)

oof_w = np.zeros_like(Y)
oof_c = np.zeros_like(Y)
fids = np.zeros(n_train, int)
skf = StratifiedGroupKFold(n_splits=N_SPLITS, shuffle=True, random_state=SEED)
for fi, (tr, va) in enumerate(skf.split(W, strata, fold_groups)):
    for j in range(len(MODS)):
        m = Ridge(alpha=1.0)
        m.fit(W[tr], Y[tr, j])
        oof_w[va, j] = m.predict(W[va])
    C = np.zeros((len(MODS), len(MODS)))
    top = Y[tr].argmax(1)
    for i in range(len(MODS)):
        sel = tr[top == i]
        C[i] = Y[sel].mean(0) if len(sel) else Y[tr].mean(0)
    t1 = oof_w[va].argmax(1)
    oof_c[va] = oof_w[va] + LAM * C[t1]
    fids[va] = fi + 1
base = ndcg5(Y, oof_w)
d = ndcg5(Y, oof_c) - base
lo, hi = np.quantile([rng.choice(d, len(d), replace=True).mean() for _ in range(2000)], [0.025, 0.975])
wf = min(ndcg5(Y[fids == k], oof_c[fids == k]).mean() - base[fids == k].mean() for k in range(1, N_SPLITS + 1))
print(f"base={base.mean():.5f} cooc={ndcg5(Y, oof_c).mean():.5f} delta={d.mean():+.5f} CI95=[{lo:+.5f},{hi:+.5f}] worst-fold={wf:+.5f}")
print("ADOPT co-occurrence boost" if (lo > GATE_FLOOR and wf >= 0) else "KEEP baseline; co-occurrence adds nothing")
```
