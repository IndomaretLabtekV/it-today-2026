# MineToday V128 — Cross-Fitted Hybrid Retrieval → Direct Top1 Utility Fusion

**Frozen experiment contract.** This notebook implements the V128 design selected after the V1–V127 audit:

- organizer data only;
- exact frozen `0.66028` champion bytes as the deployment base learner;
- exact S0/protection manifest derived from the frozen V30/V31 action ledgers;
- word/character TF-IDF + real BM25 + E5 + Nomic + fold-safe template/train-neighbor statistics + assessments;
- conservative per-module LightGBM relevance fusion;
- **direct cross-fitted exact ΔNDCG Top1 action head** (ranks 2–4 only), rather than heuristic relevance-difference authority;
- target-blind near-template prospective LOCK-P;
- DESIGN controls: full, no-BM25, no-target-neighbor, dense-only;
- multi-world validation stress tests;
- fail-close byte-identical champion output unless all preregistered gates pass.

No membership edits or B23/B34/B45 edits are allowed in V128. The final candidate only swaps existing score values between champion Top1 and one champion rank-2/3/4 module for eligible S0 users.

```python
# =========================
# 3. Frozen dense representations (E5 + Nomic), cache-provenance guarded
# =========================
E5_ID="intfloat/multilingual-e5-base"
NOMIC_ID="nomic-ai/nomic-embed-text-v2-moe"

def text_sha(texts,prefix):
    h=hashlib.sha256(); h.update(prefix.encode())
    for t in texts: h.update(b"\0"); h.update(str(t).encode("utf-8"))
    return h.hexdigest()

def _resolve_model(model_id):
    # Prefer an attached Kaggle model snapshot when internet is disabled.
    needles=[model_id.split("/")[-1].lower().replace("_","-")]
    if Path("/kaggle/input").exists():
        for p in Path("/kaggle/input").rglob("config.json"):
            s=str(p.parent).lower().replace("_","-")
            if any(n in s for n in needles): return str(p.parent)
    return model_id

def load_or_encode_e5(texts,prefix,name,device="cuda:0"):
    key=text_sha(texts,E5_ID+"|"+prefix); f=OUT/f"{name}_e5.npy"; meta=OUT/f"{name}_e5.json"
    if f.exists() and meta.exists():
        m=json.loads(meta.read_text()); arr=np.load(f)
        if m.get("key")==key and arr.shape[0]==len(texts): return np.asarray(arr,np.float32)
    import torch
    from sentence_transformers import SentenceTransformer
    mid=_resolve_model(E5_ID); model=SentenceTransformer(mid,device=device)
    arr=np.asarray(model.encode([prefix+t for t in texts],batch_size=64,show_progress_bar=True,normalize_embeddings=True,convert_to_numpy=True),np.float32)
    np.save(f,arr); meta.write_text(json.dumps({"key":key,"model":E5_ID,"resolved":mid,"prefix":prefix,"shape":list(arr.shape)},indent=2))
    del model; gc.collect(); torch.cuda.empty_cache(); return arr

def load_or_encode_nomic(texts,prefix,name,device="cuda:1"):
    key=text_sha(texts,NOMIC_ID+"|"+prefix); f=OUT/f"{name}_nomic.npy"; meta=OUT/f"{name}_nomic.json"
    if f.exists() and meta.exists():
        m=json.loads(meta.read_text()); arr=np.load(f)
        if m.get("key")==key and arr.shape[0]==len(texts): return np.asarray(arr,np.float32)
    import torch
    from transformers import AutoTokenizer,AutoModel
    mid=_resolve_model(NOMIC_ID); tok=AutoTokenizer.from_pretrained(mid,trust_remote_code=True)
    dtype=torch.float16 if torch.cuda.is_available() else torch.float32
    model=AutoModel.from_pretrained(mid,trust_remote_code=True,torch_dtype=dtype).to(device).eval()
    rows=[]
    with torch.no_grad():
        for a in range(0,len(texts),48):
            batch=[prefix+t for t in texts[a:a+48]]
            e=tok(batch,padding=True,truncation=True,max_length=256,return_tensors="pt"); e={k:v.to(device) for k,v in e.items()}
            h=model(**e).last_hidden_state; mask=e["attention_mask"].unsqueeze(-1).float()
            v=((h*mask).sum(1)/mask.sum(1).clamp_min(1)).float(); v=v/v.norm(dim=1,keepdim=True).clamp_min(1e-12)
            rows.append(v.cpu().numpy())
    arr=np.concatenate(rows).astype(np.float32); np.save(f,arr)
    meta.write_text(json.dumps({"key":key,"model":NOMIC_ID,"resolved":mid,"prefix":prefix,"shape":list(arr.shape)},indent=2))
    del model; gc.collect(); torch.cuda.empty_cache(); return arr

if not __import__('torch').cuda.is_available():
    print("WARNING: CUDA unavailable; encoding will be slow. Production target is T4x2.")
    dev_e5=dev_no="cpu"
else:
    ng=__import__('torch').cuda.device_count(); dev_e5="cuda:0"; dev_no="cuda:1" if ng>1 else "cuda:0"

E5U=load_or_encode_e5(full_text.tolist(),"query: ","users",dev_e5)
E5M=load_or_encode_e5(MOD_TEXT,"passage: ","modules",dev_e5)
NOU=load_or_encode_nomic(full_text.tolist(),"search_query: ","users",dev_no)
NOM=load_or_encode_nomic(MOD_TEXT,"search_document: ","modules",dev_no)
assert E5U.shape[0]==N and NOU.shape[0]==N and E5M.shape[0]==NOM.shape[0]==M
E5_MOD=(E5U@E5M.T).astype(np.float32); NOM_MOD=(NOU@NOM.T).astype(np.float32)
print("dense ready",E5U.shape,NOU.shape)
```

```python
# =========================
# 4. Retrieval + fold-safe target-stat feature machinery
# =========================
class BM25Index:
    def __init__(self,docs,ngram=(1,1),k1=1.2,b=0.75,max_features=70000):
        self.k1=k1; self.b=b
        self.vec=CountVectorizer(token_pattern=r"(?u)\b\w+\b",ngram_range=ngram,max_features=max_features)
        C=self.vec.fit_transform(docs).tocsr().astype(np.float32); self.n=C.shape[0]
        df=np.asarray((C>0).sum(0)).ravel(); self.idf=np.log(1+(self.n-df+0.5)/(df+0.5)).astype(np.float32)
        dl=np.asarray(C.sum(1)).ravel(); avg=max(float(dl.mean()),1e-6)
        D=C.copy()
        for i in range(self.n):
            a,bp=D.indptr[i],D.indptr[i+1]; tf=D.data[a:bp]; cols=D.indices[a:bp]
            den=tf+self.k1*(1-self.b+self.b*dl[i]/avg)
            D.data[a:bp]=(tf*(self.k1+1)/np.maximum(den,1e-12))*self.idf[cols]
        self.D=D
    def score(self,queries):
        Q=self.vec.transform(list(queries)).astype(np.float32)
        return np.asarray((Q@self.D.T).toarray(),np.float32)

# fixed user->module BM25 streams (no labels, organizer text only)
BM25_MOD_U=BM25Index(MOD_TEXT,(1,1)); BM25_MOD_B=BM25Index(MOD_TEXT,(1,2))
BM25_U_FULL=BM25_MOD_U.score(full_text); BM25_U_LATEST=BM25_MOD_U.score(latest_text)
BM25_B_FULL=BM25_MOD_B.score(full_text); BM25_B_LATEST=BM25_MOD_B.score(latest_text)

def recency_bm25(index,indices):
    out=np.zeros((len(indices),M),np.float32)
    for qi,g in enumerate(indices):
        msgs=msg_lists[g]
        if not msgs: continue
        S=index.score(msgs); w=np.power(0.5,np.arange(len(msgs)-1,-1,-1,dtype=np.float32)); w/=w.sum()
        out[qi]=(w[:,None]*S).sum(0)
    return out
BM25_U_REC=np.zeros((N,M),np.float32)
for a in range(0,N,500): BM25_U_REC[a:a+500]=recency_bm25(BM25_MOD_U,np.arange(a,min(N,a+500)))

def fit_bundle(ref_idx):
    corpus=[full_text[i] for i in ref_idx]+MOD_TEXT
    wv=TfidfVectorizer(ngram_range=(1,2),min_df=2,max_features=cfg.word_max_features,sublinear_tf=True,smooth_idf=True,norm="l2")
    cv=TfidfVectorizer(analyzer="char_wb",ngram_range=(3,5),min_df=2,max_features=cfg.char_max_features,sublinear_tf=True,norm="l2")
    wv.fit(corpus); cv.fit(corpus)
    return {
        "wv":wv,"cv":cv,
        "wfull":wv.transform(full_text.tolist()),"wlatest":wv.transform(latest_text.tolist()),"wm":wv.transform(MOD_TEXT),
        "cfull":cv.transform(full_text.tolist()),"clatest":cv.transform(latest_text.tolist()),"cm":cv.transform(MOD_TEXT)
    }

def recency_tfidf(bundle,indices):
    out=np.zeros((len(indices),M),np.float32); wv=bundle["wv"]; wm=bundle["wm"]
    for qi,g in enumerate(indices):
        msgs=msg_lists[g]
        if not msgs: continue
        S=(wv.transform(msgs)@wm.T).toarray(); w=np.power(0.5,np.arange(len(msgs)-1,-1,-1,dtype=np.float32)); w/=w.sum()
        out[qi]=(w[:,None]*S).sum(0)
    return out

def geom_from_scores(S,prefix):
    S=np.asarray(S,np.float32); order=np.argsort(-S,axis=1,kind="stable"); ranks=np.empty_like(order)
    rr=np.arange(len(S))[:,None]; ranks[rr,order]=np.arange(1,M+1)[None,:]
    best=S.max(1,keepdims=True); fifth=np.take_along_axis(S,order[:,4:5],1)
    return [(prefix+"_score",S),(prefix+"_rank",ranks.astype(np.float32)/M),(prefix+"_gapbest",best-S),(prefix+"_gap5",S-fifth)]

def impute_assessment(ref_idx,q_idx):
    med=np.nanmedian(A_RAW[ref_idx],axis=0); med=np.where(np.isfinite(med),med,2.5)
    A=np.where(np.isfinite(A_RAW[q_idx]),A_RAW[q_idx],med[None,:]).astype(np.float32)
    return A, A_BAD[q_idx].astype(np.float32)

def readiness(A,BAD):
    R=np.zeros((len(A),M),np.float32); I=np.zeros_like(R)
    for m,js in MOD_SKILLS.items():
        if js: R[:,m]=A[:,js].mean(1); I[:,m]=BAD[:,js].mean(1)
        else: R[:,m]=A.mean(1); I[:,m]=BAD.mean(1)
    return R,I

def _top_target_stats(sim,yref,k=None,scale=False):
    sim=np.asarray(sim,np.float32); nr=sim.shape[1]; k=min(cfg.knn_k if k is None else k,nr)
    prior=yref.mean(0); t1=np.eye(M,dtype=np.float32)[np.argmax(yref,1)]; p1=t1.mean(0); mem=(yref>0).astype(np.float32); pm=mem.mean(0)
    rel=np.zeros((len(sim),M),np.float32); top=np.zeros_like(rel); mm=np.zeros_like(rel)
    extra=np.zeros((len(sim),6),np.float32)
    for i,row in enumerate(sim):
        ids=np.argpartition(-row,k-1)[:k]; vals=row[ids].astype(np.float32)
        if scale:
            mx=max(float(vals.max()),1e-9); vals=vals/mx
        vals=np.clip(vals,0,None); ord=np.argsort(-vals); ids=ids[ord]; vals=vals[ord]
        w=vals*vals+1e-6; ws=float(w.sum()); ps=cfg.neighbor_prior_strength
        rel[i]=(w@yref[ids]+ps*prior)/(ws+ps); top[i]=(w@t1[ids]+ps*p1)/(ws+ps); mm[i]=(w@mem[ids]+ps*pm)/(ws+ps)
        p=w/ws; ent=float(-(p*np.log(np.maximum(p,1e-12))).sum())
        extra[i]=[vals[0],float(vals.mean()),float(vals[0]-vals[1] if len(vals)>1 else vals[0]),float((vals>=.8).sum()),float((vals>=.9).sum()),ent]
    return rel,top,mm,extra

def _group_stats(keys_ref,keys_q,yref):
    prior=yref.mean(0); p1=np.eye(M,dtype=np.float32)[np.argmax(yref,1)].mean(0); pm=(yref>0).mean(0)
    buckets=defaultdict(list)
    for i,k in enumerate(keys_ref): buckets[str(k)].append(i)
    rel=np.zeros((len(keys_q),M),np.float32); top=np.zeros_like(rel); mem=np.zeros_like(rel); ex=np.zeros((len(keys_q),2),np.float32)
    for qi,k in enumerate(keys_q):
        ids=buckets.get(str(k),[]); ps=cfg.neighbor_prior_strength
        if ids:
            yy=yref[ids]; one=np.eye(M,dtype=np.float32)[np.argmax(yy,1)]; mb=(yy>0).astype(np.float32)
            rel[qi]=(yy.sum(0)+ps*prior)/(len(ids)+ps); top[qi]=(one.sum(0)+ps*p1)/(len(ids)+ps); mem[qi]=(mb.sum(0)+ps*pm)/(len(ids)+ps)
            c=np.bincount(np.argmax(yy,1),minlength=M).astype(float); p=c/c.sum(); ent=-(p[p>0]*np.log(p[p>0])).sum(); ex[qi]=[len(ids),ent]
        else: rel[qi]=prior; top[qi]=p1; mem[qi]=pm
    return rel,top,mem,ex

def target_neighbor_features(bundle,ref_idx,q_idx):
    yref=y[ref_idx]
    sim_t=(bundle["wfull"][q_idx]@bundle["wfull"][ref_idx].T).toarray().astype(np.float32)
    bm=BM25Index([full_text[i] for i in ref_idx],(1,1)); sim_b=bm.score([full_text[i] for i in q_idx])
    sim_e=E5U[q_idx]@E5U[ref_idx].T; sim_n=NOU[q_idx]@NOU[ref_idx].T
    pieces=[]
    for name,S,scale in [("tfidf",sim_t,False),("bm25",sim_b,True),("e5",sim_e,False),("nomic",sim_n,False)]:
        r,t,m,e=_top_target_stats(S,yref,scale=scale)
        pieces += [(f"knn_{name}_rel",r),(f"knn_{name}_top1",t),(f"knn_{name}_mem",m)]
        for j,z in enumerate(["max","mean","gap","sup80","sup90","entropy"]): pieces.append((f"knn_{name}_{z}",np.repeat(e[:,j:j+1],M,1)))
    for name,kr,kq in [("exact",full_text[ref_idx],full_text[q_idx]),("skel",latest_skel[ref_idx],latest_skel[q_idx]),("near",near_group[ref_idx],near_group[q_idx])]:
        r,t,m,e=_group_stats(kr,kq,yref); pieces += [(f"tpl_{name}_rel",r),(f"tpl_{name}_top1",t),(f"tpl_{name}_mem",m),
            (f"tpl_{name}_support",np.repeat(e[:,0:1],M,1)),(f"tpl_{name}_entropy",np.repeat(e[:,1:2],M,1))]
    return pieces

def candidate_features(bundle,ref_idx,q_idx,anchor_q):
    q_idx=np.asarray(q_idx,int); pieces=[]
    wf=(bundle["wfull"][q_idx]@bundle["wm"].T).toarray(); wl=(bundle["wlatest"][q_idx]@bundle["wm"].T).toarray(); wr=recency_tfidf(bundle,q_idx)
    cf=(bundle["cfull"][q_idx]@bundle["cm"].T).toarray(); cl=(bundle["clatest"][q_idx]@bundle["cm"].T).toarray()
    for p,S in [("tfidf_word_full",wf),("tfidf_char_full",cf),("bm25_uni_full",BM25_U_FULL[q_idx]),("bm25_bi_full",BM25_B_FULL[q_idx]),("e5_mod",E5_MOD[q_idx]),("nomic_mod",NOM_MOD[q_idx])]: pieces.extend(geom_from_scores(S,p))
    pieces += [("tfidf_word_latest",wl),("tfidf_word_recency",wr),("tfidf_char_latest",cl),
               ("bm25_uni_latest",BM25_U_LATEST[q_idx]),("bm25_uni_recency",BM25_U_REC[q_idx]),("bm25_bi_latest",BM25_B_LATEST[q_idx]),
               ("dense_e5_minus_nomic",E5_MOD[q_idx]-NOM_MOD[q_idx])]
    Aq,Bq=impute_assessment(ref_idx,q_idx); R,I=readiness(Aq,Bq)
    pieces += [("ready",R),("ready_invalid",I),("assess_mean",np.repeat(Aq.mean(1,keepdims=True),M,1)),
               ("assess_std",np.repeat(Aq.std(1,keepdims=True),M,1)),("chat_count",np.repeat(chat_count[q_idx,None],M,1)),
               ("no_chat",np.repeat((chat_count[q_idx]==0).astype(np.float32)[:,None],M,1))]
    pieces.extend(geom_from_scores(anchor_q,"anchor")); pieces.extend(target_neighbor_features(bundle,ref_idx,q_idx))
    names=[]; arrays=[]
    for n,a in pieces:
        a=np.asarray(a,np.float32); assert a.shape==(len(q_idx),M),(n,a.shape)
        names.append(n); arrays.append(a[:,:,None])
    X=np.concatenate(arrays,2).astype(np.float32); assert np.isfinite(X).all()
    return X,names
```

```python
# =========================
# 5. Leakage-safe sparse anchor analogue + per-module LightGBM stage-1
# =========================
def make_inner_splits(indices,seed):
    indices=np.asarray(indices,int); groups=near_group[indices]; strat=np.argmax(y[indices],1)
    try:
        sg=StratifiedGroupKFold(cfg.n_inner,shuffle=True,random_state=seed)
        return [(indices[a],indices[b]) for a,b in sg.split(np.zeros(len(indices)),strat,groups)]
    except Exception:
        gg=GroupKFold(cfg.n_inner); return [(indices[a],indices[b]) for a,b in gg.split(np.zeros(len(indices)),groups=groups)]

def _anchor_matrix(bundle,ref_idx,q_idx,med_ref=None):
    if med_ref is None:
        med_ref=np.nanmedian(A_RAW[ref_idx],axis=0); med_ref=np.where(np.isfinite(med_ref),med_ref,2.5)
    Aq=np.where(np.isfinite(A_RAW[q_idx]),A_RAW[q_idx],med_ref[None,:]).astype(np.float32)
    B=A_BAD[q_idx].astype(np.float32); extra=np.c_[Aq,B,chat_count[q_idx],(chat_count[q_idx]==0).astype(np.float32)]
    return hstack([bundle["wfull"][q_idx],bundle["cfull"][q_idx],csr_matrix(extra)],format="csr")

def nested_anchor(bundle,train_idx,query_idx,inner_splits):
    train_idx=np.asarray(train_idx,int); pos={g:i for i,g in enumerate(train_idx)}
    oof_rel=np.zeros((len(train_idx),M),np.float32); oof_top=np.zeros_like(oof_rel)
    for f,(tr,va) in enumerate(inner_splits):
        med=np.nanmedian(A_RAW[tr],axis=0); med=np.where(np.isfinite(med),med,2.5)
        Xtr=_anchor_matrix(bundle,tr,tr,med); Xv=_anchor_matrix(bundle,tr,va,med)
        r=Ridge(alpha=3.0,solver="lsqr",tol=2e-3,max_iter=1000).fit(Xtr,y[tr])
        t=Ridge(alpha=3.0,solver="lsqr",tol=2e-3,max_iter=1000).fit(Xtr,np.eye(M,dtype=np.float32)[np.argmax(y[tr],1)])
        loc=np.array([pos[g] for g in va]); oof_rel[loc]=r.predict(Xv); oof_top[loc]=t.predict(Xv)
    med=np.nanmedian(A_RAW[train_idx],axis=0); med=np.where(np.isfinite(med),med,2.5)
    Xtr=_anchor_matrix(bundle,train_idx,train_idx,med); Xq=_anchor_matrix(bundle,train_idx,query_idx,med)
    r=Ridge(alpha=3.0,solver="lsqr",tol=2e-3,max_iter=1000).fit(Xtr,y[train_idx])
    t=Ridge(alpha=3.0,solver="lsqr",tol=2e-3,max_iter=1000).fit(Xtr,np.eye(M,dtype=np.float32)[np.argmax(y[train_idx],1)])
    oof=(.9*oof_rel+.1*oof_top).astype(np.float32); q=(.9*r.predict(Xq)+.1*t.predict(Xq)).astype(np.float32)
    return oof,q

def variant_columns(names,variant):
    if variant=="full": return np.arange(len(names))
    if variant=="no_bm25": return np.array([i for i,n in enumerate(names) if "bm25" not in n],int)
    if variant=="no_neighbor": return np.array([i for i,n in enumerate(names) if not (n.startswith("knn_") or n.startswith("tpl_"))],int)
    if variant=="dense_only":
        keep=("e5_","nomic_","dense_","ready","assess_","chat_","no_chat","anchor_")
        return np.array([i for i,n in enumerate(names) if n.startswith(keep)],int)
    raise ValueError(variant)

def fit_lgb_variant(Xtr,ytr,Xq,names,variant):
    cols=variant_columns(names,variant); pred=np.zeros((len(Xq),M),np.float32)
    for m in range(M):
        vals=[]
        for seed in cfg.lgb_seeds:
            model=lgb.LGBMRegressor(objective="regression_l2",n_estimators=cfg.lgb_estimators,learning_rate=cfg.lgb_lr,
                num_leaves=cfg.lgb_num_leaves,max_depth=cfg.lgb_max_depth,min_child_samples=cfg.lgb_min_child,
                subsample=cfg.lgb_subsample,subsample_freq=1,colsample_bytree=cfg.lgb_colsample,reg_alpha=cfg.lgb_reg_alpha,
                reg_lambda=cfg.lgb_reg_lambda,random_state=int(seed+37*m),n_jobs=2,verbosity=-1,force_col_wise=True)
            model.fit(Xtr[:,m,cols],ytr[:,m]); vals.append(model.predict(Xq[:,m,cols]))
        pred[:,m]=np.mean(vals,axis=0)
    return pred

def train_query_stage1(train_idx,query_idx,variants,seed=SEED):
    train_idx=np.asarray(train_idx,int); query_idx=np.asarray(query_idx,int); bundle=fit_bundle(train_idx)
    inner=make_inner_splits(train_idx,seed+11); anchor_tr,anchor_q=nested_anchor(bundle,train_idx,query_idx,inner)
    Xtr=None; names=None; pos={g:i for i,g in enumerate(train_idx)}
    for fi,(ir,iv) in enumerate(inner):
        loc=np.array([pos[g] for g in iv]); Xi,ni=candidate_features(bundle,ir,iv,anchor_tr[loc])
        if Xtr is None: Xtr=np.zeros((len(train_idx),M,Xi.shape[2]),np.float32); names=ni
        assert ni==names; Xtr[loc]=Xi
    Xq,nq=candidate_features(bundle,train_idx,query_idx,anchor_q); assert nq==names and np.isfinite(Xtr).all()
    preds={v:fit_lgb_variant(Xtr,y[train_idx],Xq,names,v) for v in variants}
    gc.collect()
    return {"anchor_train":anchor_tr,"anchor_query":anchor_q,"Xtrain":Xtr,"Xquery":Xq,"names":names,"query":preds}

def make_world_splits(indices,world,seed=SEED):
    indices=np.asarray(indices,int); strat=np.array([f"{np.argmax(y[i])}_{int((y[i]>0).sum())}" for i in indices],object)
    # collapse rare strata only for stratification; grouping remains unchanged.
    cc=Counter(strat); strat=np.array([s if cc[s]>=cfg.n_outer else str(np.argmax(y[i])) for s,i in zip(strat,indices)],object)
    if world=="random":
        sk=StratifiedKFold(cfg.n_outer,shuffle=True,random_state=seed); return [(indices[a],indices[b]) for a,b in sk.split(np.zeros(len(indices)),strat)]
    if world in {"exact","latest","near"}:
        groups={"exact":full_text[indices],"latest":latest_skel[indices],"near":near_group[indices]}[world]
        try:
            sg=StratifiedGroupKFold(cfg.n_outer,shuffle=True,random_state=seed); return [(indices[a],indices[b]) for a,b in sg.split(np.zeros(len(indices)),strat,groups)]
        except Exception:
            gg=GroupKFold(cfg.n_outer); return [(indices[a],indices[b]) for a,b in gg.split(np.zeros(len(indices)),groups=groups)]
    if world=="contiguous":
        order=indices[np.argsort([int(re.sub(r"\D","",ALL_USERS[i])) for i in indices])]; blocks=np.array_split(order,cfg.n_outer)
        return [(np.setdiff1d(indices,b),np.asarray(b,int)) for b in blocks]
    raise ValueError(world)

def oof_stage1(indices,world,variants):
    indices=np.asarray(indices,int); local={g:i for i,g in enumerate(indices)}
    wseed=cfg.seed+int(hashlib.sha1(world.encode()).hexdigest()[:6],16)%10000
    splits=make_world_splits(indices,world,wseed)
    anchor=np.zeros((len(indices),M),np.float32); fusion={v:np.zeros_like(anchor) for v in variants}; raw=None; names=None; fold_id=np.full(len(indices),-1,int)
    rows=[]
    for f,(tr,va) in enumerate(splits):
        t=time.time(); z=train_query_stage1(tr,va,variants,cfg.seed+1000*f)
        lv=np.array([local[g] for g in va]); anchor[lv]=z["anchor_query"]; fold_id[lv]=f
        if raw is None: raw=np.zeros((len(indices),M,z["Xquery"].shape[2]),np.float32); names=z["names"]
        assert z["names"]==names; raw[lv]=z["Xquery"]
        for v in variants: fusion[v][lv]=z["query"][v]
        rows.append({"world":world,"fold":f,"n":len(va),"anchor_ndcg":ndcg5(y[va],z["anchor_query"]),"seconds":time.time()-t})
        del z; gc.collect()
    assert (fold_id>=0).all() and np.isfinite(raw).all()
    return {"indices":indices,"anchor":anchor,"fusion":fusion,"raw":raw,"names":names,"fold_id":fold_id,"folds":pd.DataFrame(rows)}
```

```python
# =========================
# 6. Direct exact-ΔNDCG action head + risk/coverage policy
# =========================
def action_table(indices,anchor,fusion,raw,names):
    n=len(indices); F=raw.shape[2]; X=[]; Y=[]; users=[]; cand=[]; inc=[]; ranks=[]
    order=np.argsort(-anchor,axis=1,kind="stable")
    for i,g in enumerate(indices):
        im=int(order[i,0])
        for rr in (1,2,3):
            cm=int(order[i,rr]); s=anchor[i].copy()
            if g < NTR:
                before=per_user_ndcg(y[g:g+1],s[None,:])[0]
                s[im],s[cm]=s[cm],s[im]
                delta=float(per_user_ndcg(y[g:g+1],s[None,:])[0]-before)
            else:
                delta=0.0
            diff=raw[i,cm]-raw[i,im]
            pair=np.zeros(M*M,np.float32); pair[im*M+cm]=1
            c1=np.zeros(M,np.float32); c1[cm]=1; i1=np.zeros(M,np.float32); i1[im]=1; r1=np.zeros(3,np.float32); r1[rr-1]=1
            feat=np.r_[fusion[i,cm],fusion[i,im],fusion[i,cm]-fusion[i,im],anchor[i,im]-anchor[i,cm],diff,c1,i1,pair,r1]
            X.append(feat); Y.append(delta); users.append(i); cand.append(cm); inc.append(im); ranks.append(rr+1)
    fn=["fusion_c","fusion_i","fusion_diff","anchor_margin"]+["d_"+x for x in names]+[f"cand_{m}" for m in MODULES]+[f"inc_{m}" for m in MODULES]+[f"pair_{a}_{b}" for a in MODULES for b in MODULES]+["rank2","rank3","rank4"]
    return np.asarray(X,np.float32),np.asarray(Y,np.float32),np.asarray(users,int),np.asarray(cand,int),np.asarray(inc,int),np.asarray(ranks,int),fn

def action_oof(stage,variant):
    X,Y,U,C,I,R,fn=action_table(stage["indices"],stage["anchor"],stage["fusion"][variant],stage["raw"],stage["names"])
    pred=np.zeros(len(X),np.float32)
    for f in range(cfg.n_outer):
        vu=np.flatnonzero(stage["fold_id"]==f); trmask=~np.isin(U,vu); vamask=np.isin(U,vu)
        model=make_pipeline(StandardScaler(),Ridge(alpha=cfg.action_alpha,solver="lsqr",tol=1e-5,max_iter=3000))
        model.fit(X[trmask],Y[trmask]); pred[vamask]=model.predict(X[vamask])
    return {"X":X,"y":Y,"user":U,"cand":C,"inc":I,"rank":R,"pred":pred,"features":fn}

def bootstrap_ci(d,groups=None,seed=SEED):
    d=np.asarray(d,float); rng=np.random.default_rng(seed); vals=np.empty(cfg.bootstraps,float)
    if groups is None:
        n=len(d)
        for b in range(cfg.bootstraps): vals[b]=d[rng.integers(0,n,n)].mean()
    else:
        groups=np.asarray(groups); ug=np.unique(groups); bucket={g:np.flatnonzero(groups==g) for g in ug}
        for b in range(cfg.bootstraps):
            pick=rng.choice(ug,len(ug),replace=True); ix=np.concatenate([bucket[g] for g in pick]); vals[b]=d[ix].mean()
    return float(np.quantile(vals,.025)),float(np.quantile(vals,.975))

def policy_at_coverage(act,n_users,coverage,groups=None,fold_id=None):
    P=act["pred"].reshape(n_users,3); D=act["y"].reshape(n_users,3); C=act["cand"].reshape(n_users,3)
    j=np.argmax(P,1); best=P[np.arange(n_users),j]; realized=D[np.arange(n_users),j]; bc=C[np.arange(n_users),j]
    k=max(1,int(math.ceil(coverage*n_users))); elig=np.flatnonzero(best>0); chosen=elig[np.argsort(-best[elig])[:min(k,len(elig))]]
    ud=np.zeros(n_users,float); ud[chosen]=realized[chosen]; sel=np.zeros(n_users,bool); sel[chosen]=True
    ciu=bootstrap_ci(ud,seed=SEED+11); cig=bootstrap_ci(ud,groups,SEED+12) if groups is not None else ciu
    out={"coverage":float(sel.mean()),"requested":float(coverage),"gain":float(ud.mean()),"n":int(sel.sum()),"benefit":int((ud>0).sum()),"harm":int((ud<0).sum()),
         "tie":int(sel.sum()-(ud!=0).sum()),"ci_user":ciu,"ci_cluster":cig,"user_delta":ud,"selected":sel,"candidate":bc,"score":best}
    if fold_id is not None:
        fg=[]
        for f in sorted(np.unique(fold_id)):
            mask=fold_id==f; fg.append(float(ud[mask].mean()))
        out["fold_gains"]=fg
    return out

def choose_policy(act,n_users,groups,fold_id):
    rows=[]; objs=[]
    for c in cfg.coverage_grid:
        z=policy_at_coverage(act,n_users,c,groups,fold_id); objs.append(z)
        rows.append({k:z[k] for k in ["requested","coverage","gain","n","benefit","harm","tie"]}|{"ci_user_low":z["ci_user"][0],"ci_user_high":z["ci_user"][1],"ci_cluster_low":z["ci_cluster"][0],"ci_cluster_high":z["ci_cluster"][1]})
    valid=[z for z in objs if z["gain"]>0 and z["benefit"]>=2*max(1,z["harm"])]
    if not valid: valid=[max(objs,key=lambda z:z["gain"])]
    best=max(valid,key=lambda z:z["gain"]); se=float(np.std(best["user_delta"],ddof=1)/math.sqrt(n_users))
    near=[z for z in valid if z["gain"]>=best["gain"]-se]; chosen=min(near,key=lambda z:z["requested"])
    return chosen,pd.DataFrame(rows)

def fit_action_model(act):
    model=make_pipeline(StandardScaler(),Ridge(alpha=cfg.action_alpha,solver="lsqr",tol=1e-5,max_iter=3000)); model.fit(act["X"],act["y"]); return model

def predict_action(model,indices,anchor,fusion,raw,names):
    X,Y,U,C,I,R,fn=action_table(indices,anchor,fusion,raw,names); return {"X":X,"y":Y,"user":U,"cand":C,"inc":I,"rank":R,"pred":model.predict(X).astype(np.float32),"features":fn}

def policy_predict(act,n_users,coverage,eligible=None):
    P=act["pred"].reshape(n_users,3); C=act["cand"].reshape(n_users,3); I=act["inc"].reshape(n_users,3)
    j=np.argmax(P,1); score=P[np.arange(n_users),j]; cand=C[np.arange(n_users),j]; inc=I[np.arange(n_users),j]
    ok=score>0
    if eligible is not None: ok &= np.asarray(eligible,bool)
    ids=np.flatnonzero(ok); k=max(1,int(math.ceil(coverage*n_users))); pick=ids[np.argsort(-score[ids])[:min(k,len(ids))]]
    selected=np.zeros(n_users,bool); selected[pick]=True
    return selected,cand,inc,score
```

```python
# =========================
# 7. DESIGN primary world: full/no-BM25/no-neighbor/dense-only + frozen branch selection
# =========================
PRIMARY_WORLD="near"; VARIANTS=["full","no_bm25","no_neighbor","dense_only"]
t0=time.time(); primary=oof_stage1(DESIGN,PRIMARY_WORLD,VARIANTS)
control_rows=[]; action_objs={}; policies={}; curves={}
for v in VARIANTS:
    a=action_oof(primary,v); p,curve=choose_policy(a,len(DESIGN),near_group[DESIGN],primary["fold_id"])
    action_objs[v]=a; policies[v]=p; curves[v]=curve.assign(variant=v)
    control_rows.append({"variant":v,"gain":p["gain"],"coverage":p["coverage"],"n":p["n"],"benefit":p["benefit"],"harm":p["harm"],
                         "ci_user_low":p["ci_user"][0],"ci_cluster_low":p["ci_cluster"][0],"worst_fold":min(p.get("fold_gains",[0]))})
controls=pd.DataFrame(control_rows).sort_values("gain",ascending=False); display(controls)
pd.concat(curves.values(),ignore_index=True).to_csv(OUT/"v128_risk_coverage_design.csv",index=False)
controls.to_csv(OUT/"v128_design_control_metrics.csv",index=False)

# BM25 is retained only if it adds stable matched-coverage utility. Otherwise freeze no-BM25 before LOCK-P.
full=policies["full"]; nb_match=policy_at_coverage(action_objs["no_bm25"],len(DESIGN),full["requested"],near_group[DESIGN],primary["fold_id"])
bm25_fold=np.asarray(full["fold_gains"])-np.asarray(nb_match["fold_gains"])
bm25_stable=bool(full["gain"]>=nb_match["gain"] and (bm25_fold>0).sum()>=3)
SELECTED="full" if bm25_stable else "no_bm25"
selected=policies[SELECTED]; cov=float(selected["requested"])
dense_match=policy_at_coverage(action_objs["dense_only"],len(DESIGN),cov,near_group[DESIGN],primary["fold_id"])
neighbor_match=policy_at_coverage(action_objs["no_neighbor"],len(DESIGN),cov,near_group[DESIGN],primary["fold_id"])
DESIGN_CONTROL_OK=bool(selected["gain"]>dense_match["gain"] and selected["gain"]>0 and selected["ci_user"][0]>0 and selected["ci_cluster"][0]>0)
print("SELECTED",SELECTED,"coverage",cov,"design gain",selected["gain"],"BM25 stable",bm25_stable,"design control gate",DESIGN_CONTROL_OK)

# Exact design action ledger.
P=action_objs[SELECTED]["pred"].reshape(len(DESIGN),3); C=action_objs[SELECTED]["cand"].reshape(len(DESIGN),3); D=action_objs[SELECTED]["y"].reshape(len(DESIGN),3)
j=np.argmax(P,1); ledger=pd.DataFrame({"user_id":train.user_id.iloc[DESIGN].astype(str).to_numpy(),"selected":selected["selected"],"candidate":np.array(MODULES)[C[np.arange(len(DESIGN)),j]],
    "pred_utility":P[np.arange(len(DESIGN)),j],"realized_delta":selected["user_delta"],"near_group":near_group[DESIGN],"fold":primary["fold_id"]})
ledger.to_csv(OUT/"v128_action_ledger_design.csv",index=False)
json.dump(primary["names"],open(OUT/"v128_feature_names.json","w"),indent=2)
print("primary DESIGN completed in",round(time.time()-t0,1),"s")
```

```python
# =========================
# 8. DESIGN multi-world robustness at the frozen selected coverage
# =========================
world_rows=[]
# primary world result first
world_rows.append({"world":"near","gain":selected["gain"],"coverage":selected["coverage"],"n":selected["n"],"worst_fold":min(selected["fold_gains"]),
                   "ci_user_low":selected["ci_user"][0],"ci_cluster_low":selected["ci_cluster"][0]})
for world in ["random","exact","latest","contiguous"]:
    st=oof_stage1(DESIGN,world,[SELECTED]); ac=action_oof(st,SELECTED); po=policy_at_coverage(ac,len(DESIGN),cov,near_group[DESIGN],st["fold_id"])
    world_rows.append({"world":world,"gain":po["gain"],"coverage":po["coverage"],"n":po["n"],"worst_fold":min(po["fold_gains"]),
                       "ci_user_low":po["ci_user"][0],"ci_cluster_low":po["ci_cluster"][0]})
    del st,ac; gc.collect()
worlds=pd.DataFrame(world_rows); display(worlds); worlds.to_csv(OUT/"v128_world_metrics.csv",index=False)
WORLD_GATE=bool((worlds.gain>=0).all() and (worlds.worst_fold>=-0.00025).all())
print("WORLD_GATE",WORLD_GATE)
```

```python
# =========================
# 9. One-shot LOCK-P. No model/coverage/variant changes after this cell begins.
# =========================
# Train DESIGN -> LOCK with exactly the frozen SELECTED branch.
t0=time.time(); zlock=train_query_stage1(DESIGN,LOCK,[SELECTED],cfg.seed+70000)
# Action head is trained only on cross-fitted DESIGN action data selected above.
action_model=fit_action_model(action_objs[SELECTED])
lock_act=predict_action(action_model,LOCK,zlock["anchor_query"],zlock["query"][SELECTED],zlock["Xquery"],zlock["names"])
sel_l,cand_l,inc_l,score_l=policy_predict(lock_act,len(LOCK),cov)
# Realized exact lock deltas are evaluated only now.
Dlock=lock_act["y"].reshape(len(LOCK),3); Plock=lock_act["pred"].reshape(len(LOCK),3); jl=np.argmax(Plock,1)
ud=np.zeros(len(LOCK),float); ud[sel_l]=Dlock[np.arange(len(LOCK)),jl][sel_l]
ciu=bootstrap_ci(ud,seed=cfg.seed+71); cic=bootstrap_ci(ud,near_group[LOCK],cfg.seed+72)
seen=np.isin(latest_skel[LOCK],np.unique(latest_skel[DESIGN])); gs=float(ud[seen].mean()) if seen.any() else 0.0; gu=float(ud[~seen].mean()) if (~seen).any() else 0.0
G_LOCK=float(ud.mean()); G_MIX=.816*gs+.184*gu; H=max((ciu[1]-ciu[0])/2,(cic[1]-cic[0])/2); S=max(0.0,float(selected["gain"])-G_LOCK); G_SAFE=min(G_LOCK,G_MIX)-H-S
benef=int((ud>0).sum()); harm=int((ud<0).sum())
# Module blocks by anchor Top1 module, supported n>=100.
mods=np.argmax(zlock["anchor_query"],1); block=[]
for m in range(M):
    mask=mods==m
    if mask.sum(): block.append({"module":MODULES[m],"n":int(mask.sum()),"gain":float(ud[mask].mean())})
blockdf=pd.DataFrame(block); blockdf.to_csv(OUT/"v128_lock_module_blocks.csv",index=False)
block_ok=bool((blockdf.loc[blockdf.n>=100,"gain"]>=-0.001).all()) if len(blockdf) else True
LOCK_GATE=bool(G_LOCK>0 and ciu[0]>0 and cic[0]>0 and G_SAFE>=cfg.top5_gap and block_ok)
lock_report={"selected_variant":SELECTED,"coverage":cov,"gain":G_LOCK,"seen_gain":gs,"unseen_gain":gu,"mix_gain":G_MIX,"ci_user":ciu,"ci_cluster":cic,
             "uncertainty_halfwidth":H,"selection_optimism":S,"g_safe":G_SAFE,"n":int(sel_l.sum()),"benefit":benef,"harm":harm,"module_blocks_ok":block_ok,"lock_gate":LOCK_GATE}
json.dump(lock_report,open(OUT/"v128_lock_metrics.json","w"),indent=2); print(json.dumps(lock_report,indent=2))
lock_ledger=pd.DataFrame({"user_id":train.user_id.iloc[LOCK].astype(str).to_numpy(),"selected":sel_l,"candidate":np.array(MODULES)[cand_l],"pred_utility":score_l,"realized_delta":ud,"seen_latest_skeleton":seen,"near_group":near_group[LOCK]})
lock_ledger.to_csv(OUT/"v128_action_ledger_lock.csv",index=False)
print("LOCK-P evaluated once in",round(time.time()-t0,1),"s")
```

```python
# =========================
# 10. Final all-train refit -> exact champion Top1 score-swap candidate -> fail-close primary
# =========================
PRE_GATE=bool(DESIGN_CONTROL_OK and WORLD_GATE and LOCK_GATE and not cfg.smoke)
print("PRE_GATE",PRE_GATE)
# Generate final model only when the prospective gate is scientifically open; otherwise save time and byte-copy champion.
candidate_path=OUT/"candidate_v128_DIAGNOSTIC.csv"; primary_path=OUT/"submission_v128_primary.csv"
TEST_GEOM_GATE=False; test_actions=0; test_rate=0.0
if PRE_GATE:
    ztest=train_query_stage1(ALL_TRAIN,np.arange(NTR,N),[SELECTED],cfg.seed+90000)
    # Keep the prospectively frozen DESIGN action head for deployment. Only stage-1 relevance is refit on all 4,000 labels.
    # This preserves action-score scale/coverage and avoids a second-order OOF stacking path after LOCK-P.
    final_model=action_model
    # Deployment candidate set is defined by the exact champion ranks 2-4, not the anchor analogue ranks.
    test_indices=np.arange(NTR,N)
    test_act=predict_action(final_model,test_indices,CHAMPION_SCORES,ztest["query"][SELECTED],ztest["Xquery"],ztest["names"])
    sel,cand,inc,score=policy_predict(test_act,NTE,cov,TEST_ELIGIBLE)
    cand_scores=CHAMPION_SCORES.copy()
    champ_order=np.argsort(-CHAMPION_SCORES,axis=1,kind="stable")
    # predict_action was built with champion anchor, so inc/cand are guaranteed to be champion Top1/rank2-4.
    for r in np.flatnonzero(sel):
        a=int(inc[r]); b=int(cand[r]); assert a==int(champ_order[r,0]) and b in set(map(int,champ_order[r,1:4]))
        cand_scores[r,a],cand_scores[r,b]=cand_scores[r,b],cand_scores[r,a]
    out=champion.copy(); out.loc[:,MODULES]=cand_scores; out.to_csv(candidate_path,index=False)
    test_actions=int(sel.sum()); test_rate=float(test_actions/NTE)
    # Geometry/protection assertions.
    ta,tb=top5_idx(CHAMPION_SCORES),top5_idx(cand_scores); changed=np.any(ta!=tb,axis=1)
    membership=np.array([set(a)!=set(b) for a,b in zip(ta,tb)])
    locked=(~TEST_ELIGIBLE)
    exact_locked=bool(np.array_equal(cand_scores[locked],CHAMPION_SCORES[locked]))
    rank5_same=bool(np.array_equal(ta[:,4],tb[:,4])); multiset_ok=True
    for r in np.flatnonzero(changed):
        if not np.array_equal(np.sort(CHAMPION_SCORES[r]),np.sort(cand_scores[r])): multiset_ok=False; break
    # 95% interval of DESIGN outer-fold action-rate behavior at frozen coverage.
    rates=[]
    for f in range(cfg.n_outer):
        m=primary["fold_id"]==f; rates.append(float(selected["selected"][m].mean()))
    rates=np.asarray(rates); lo=max(0,float(rates.mean()-1.96*rates.std(ddof=1)/math.sqrt(len(rates)))); hi=min(1,float(rates.mean()+1.96*rates.std(ddof=1)/math.sqrt(len(rates))))
    rate_ok=bool(lo-1e-9<=test_rate<=hi+1e-9)
    TEST_GEOM_GATE=bool(exact_locked and not membership.any() and rank5_same and multiset_ok and rate_ok and test_actions<=math.ceil(max(cfg.coverage_grid)*NTE))
    test_ledger=pd.DataFrame({"user_id":test.user_id.astype(str),"eligible_s0":TEST_ELIGIBLE,"selected":sel,"candidate":np.array(MODULES)[cand],"incumbent":np.array(MODULES)[inc],"pred_utility":score})
    test_ledger.to_csv(OUT/"v128_action_ledger_test.csv",index=False)
else:
    candidate_path.write_bytes(CHAMPION_BYTES)

FINAL_GATE=bool(PRE_GATE and TEST_GEOM_GATE)
if FINAL_GATE:
    primary_path.write_bytes(candidate_path.read_bytes())
else:
    primary_path.write_bytes(CHAMPION_BYTES)
    assert hashlib.sha256(primary_path.read_bytes()).hexdigest()==CHAMPION_SHA256

pkg={"python":platform.python_version(),"numpy":np.__version__,"pandas":pd.__version__,"scipy":scipy.__version__,"sklearn":__import__('sklearn').__version__,"lightgbm":lgb.__version__}
json.dump(pkg,open(OUT/"v128_package_versions.json","w"),indent=2)
report={"version":"V128","selected_variant":SELECTED,"coverage":cov,"design_gain":float(selected["gain"]),"design_ci_user":selected["ci_user"],"design_ci_cluster":selected["ci_cluster"],
        "bm25_stable":bm25_stable,"design_control_gate":DESIGN_CONTROL_OK,"world_gate":WORLD_GATE,"lock_gate":LOCK_GATE,"g_safe":G_SAFE,
        "test_geometry_gate":TEST_GEOM_GATE,"final_gate":FINAL_GATE,"test_actions":test_actions,"test_action_rate":test_rate,
        "champion_sha256":CHAMPION_SHA256,"primary_sha256":hashlib.sha256(primary_path.read_bytes()).hexdigest(),
        "submission_instruction":"SUBMIT submission_v128_primary.csv" if FINAL_GATE else "NO V128 SUBMISSION: primary is byte-identical 0.66028 champion."}
json.dump(report,open(OUT/"v128_gate_report.json","w"),indent=2)
print(json.dumps(report,indent=2)); print("Outputs:",OUT)
```

## Expected output artifacts

The notebook always writes `v128/submission_v128_primary.csv`. If any DESIGN, multi-world, LOCK-P, provenance, or geometry gate fails, that file is **byte-identical** to the frozen `0.66028` champion and should not consume a submission slot.

If the final gate passes, the same directory contains the immutable action ledger, validation/control tables, LOCK metrics, feature names, package versions, and a diagnostic candidate CSV. Do not manually edit users after observing the public leaderboard.
